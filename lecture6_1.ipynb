```python
#wap to access each character of string in forward and backward direction by using while loop
#i/p= "learning python is very easy"

s="learning python is very easy"
n=len(s)
i=0
print("forward direction")
while i<n:
    print(s[i],end= ' ')
    i+=1
print("backward direction")
i=1
while i>=-n:
    print(s[i],end=' ')
    i=i-1
```

```python
#insertion sort
arr=[5,3,8,6,2]
for i in range (1,len(arr)):
    key=arr[i]
    j=i-1
    while j>=0 and arr[j]>key:
        arr[j+1]=arr[j]
        j=j-1
    arr[j+1]=key
    print(arr)
```

```python
# Selection Sort
arr1 = [20, 12, 10, 15, 2]

for i in range(len(arr1)):
    min_index = i
    j = i + 1

    while j < len(arr1):
        if arr1[j] < arr1[min_index]:
            min_index = j
        j += 1

    arr1[i], arr1[min_index] = arr1[min_index], arr1[i]
    print(arr1)

print("Sorted array:", arr1)
```
