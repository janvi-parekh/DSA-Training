```python
#recurssion: when the main problem can be divided into the similar sub problems

def factorial(num):
    if num<=1:
        return 1
    return num * factorial(num-1)
print(factorial(4))
```

```python
#capatalizeFirst solution using recurssion
def capitalizeFirst(arr):
    result =[]
    if len(arr)==0:
        return result
    result.append(arr[0][0].upper()+arr[0][1:])
    return result+capitalizeFirst(arr[1:])

print(capitalizeFirst(["car","taco","banana"]))
```

```python
#power and exponent
def power(base,exponent):
    if exponent ==0:
        return 1
    return base*power(base,exponent-1)
print(power(2,0))
print(power(2,2))
print(power(2,4))
```

```python
#multiply elements of array
def productofArray(arr):
    if len(arr)==0:
        return 1
    return arr[0]* productofArray(arr[1:])
print(productofArray([1,2,3]))
print(productofArray([1,2,3,10]))        
```

```python
#reverse a string
def reverse(str):
    if len(str) <=1:
        return str
    return str[len(str)-1]+reverse(str[0:len(str)-1])
print(reverse("python"))
```

```python
def recursiverange(num):
    if num <=0:
        return 0
    return num + recursiverange(num-1)
print(recursiverange(6))
```

```python
def isPalindrome(str):
    if len(str)==0:
        return True
    if str[0] != str[len(str)-1]:
        return False
    return isPalindrome(str[1:-1]) # string will be sliced form 1 to -2
    
print(isPalindrome("madam"))
```
