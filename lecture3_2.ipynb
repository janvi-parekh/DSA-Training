```python
#linear search

def linearSearch(array,target):
    for i in range(0,len(array)):
        if array[i] == target:
            return i

array=[1,2,3,4,8,9,7]
target=10
result = linearSearch(array, target)
if result == -1:
    print("target not found")
else:
    print("target found")
```
