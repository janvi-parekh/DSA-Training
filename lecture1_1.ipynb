```python
#for 
for i in range (5): #by default starts from 0 if starting point of the rage is not defined, no increment operator needed, because it internally sequencially increments
    print(i)
   
print("from defined starting ")
for i in range (2,10):
    print(i)

print("from defined starting and difference bwt values eg increment")
for i in range (2,10,2):
    print(i)

print("from defined starting and difference bwt values eg decrement")
for i in range (10,0,-2):
    print(i)
```

```python
#break and continue

for i in range (1,6):
    if i==3:
        break
    print(i)
    

for i in range(1,6):
    if i==3:
        continue
    print(i)
```

```python
#zip function
for i,j in zip(range(1,6), range(5,0,-1)):
    if i==3 and j==3:
        continue
    print (i,"  ",j)
```
