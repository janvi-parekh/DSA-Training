```python
name="madam"
newname =""
N=len(name)
for i in range(N-1,-1,-1):
        newname+=name[i]
print(newname)

if name==newname:
    print("Palindrom")
else:
    print("Not Palindrom")
```

```python
import time
n=int(input("Enter the number of rows: "))
for i in range(1,n+1):
    print(" "*(n-i),end=" ")
    for j in range(1,i+1):
        time.sleep(3)
        print("*",end=" ")
    print()
```
