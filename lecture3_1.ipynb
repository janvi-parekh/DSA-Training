```python
# how many type of arguments we can pass in functin? 
# 1. Positional argument : argument positions are fixed from left to right. 
def arithematic(a,b):
    sum = a+b
    sub = a-b
    mul=a*b
    div=a/b
    return sum,sub,div,mul
print(arithematic(5,5))
```

```python
#2. Keyword argument : keyword name and parameter name should be same
def credential(user, pwd):
    if user == pwd:
        print("login successful")
    else:
        print("invalid credential")
credential(user="admin",pwd="admin")
```

```python
#3.Default argument: id no value is passed in variable the the function will take default value
def cityname(city="Pune"):
    print(city)
cityname("Nagpur")
cityname("Mumbai")
cityname()
```

```python
#4. variable length argument : selects all argument values and store it in tuple, use * for selecting all values
def cityname(*name):
    print(name)
cityname("nagpur","delhi","Mumbai")
```

```python
import sys
def add():
    a= int(input("Enter first value: "))
    b= int(input("Enter second value: "))
    print(a+b)

def sub():
    a= int(input("Enter first value: "))
    b= int(input("Enter second value: "))
    print(a-b)

def div():
    a= int(input("Enter first value: "))
    b= int(input("Enter second value: "))
    print(a/b)

def mul():
    a= int(input("Enter first value: "))
    b= int(input("Enter second value: "))
    print(a*b)

while True:
    print("1.Addition")
    print("2.suntraction")
    print("3.division")
    print("4.multiplition")
    print("5.exit")
    choice =int(input("Enter your choice"))
    if choice == 1:
        add()
    elif choice ==2:
        sub()
    elif choice ==3:
        div()
    elif choice ==4:
        mul()
    elif choice==5:
        sys.exit()
```

```python
#complexicity  (time complexicity)
def findbiggestnumber(samplearray):
    biggestnumber=samplearray[0]
    for index in range(1,len(samplearray)):
        if samplearray[index]>biggestnumber:
            biggestnumber=samplearray[index]
    print(biggestnumber)
samplearray=[5,7,9,2,3,4]
findbiggestnumber(samplearray)
```
