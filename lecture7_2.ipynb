```python
class Tree:
    def __init__(self,data):
        self.data=data
        self.child=[]

    def __str__(self,level=0):
        ret=" " *level+str(self.data)+"\n"
        for ch in self.child:
            ret +=ch.__str__(level+1)
        return ret

    def addChild(self,obj):
        self.child.append(obj)
        print("Tree node added")

rootNode= Tree("Drinks")
hot=Tree("Hot")
cold=Tree("Cold")
tea=Tree("Tea")
coffee=Tree("Coffee")
icetea=Tree("IceTea")
coldcoffee=Tree("Cold Coffee")

rootNode.addChild(hot)
rootNode.addChild(cold)
hot.addChild(tea)
hot.addChild(coffee)
cold.addChild(icetea)
cold.addChild(coldcoffee)

print(rootNode)
```
